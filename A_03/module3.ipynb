```python
import numpy as np
from module3 import and_gate, nand_gate, nor_gate, or_gate, xor_gate

# numpy array of all four input combinations for the logic gates
# [0, 0]
# [0, 1] 
# [1, 0] 
# [1, 1]
inputs = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
```

# AND Gate

```python
# loop through the input array and compute the outputs for the AND gate and save them in an array
and_results = np.array([and_gate.compute(row) for row in inputs])

print("AND Gate Results:")
print(and_results)
```
