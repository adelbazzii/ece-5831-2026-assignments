Working with files

```python
with open('readme.txt') as f:
    lines = f.readlines()
print(lines)
```

```python
with open('readme.txt') as f:
    while True:
        line = f.readline()
        if not line:
            break
        print(line.strip())
```

```python
with open('readme.txt') as f:
    for line in f:
        print(line.strip())
```

```python
with open('readme.txt', 'w') as f:
    f.write('readme')
```

```python
import os.path

file_exists = os.path.exists('readme.txt')

print(file_exists)
```

```python
import os

os.rename('readme.txt', 'notes.txt')
```
